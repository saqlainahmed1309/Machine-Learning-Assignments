```python
import pandas as pd
import numpy as np
```

# Load the dataset

```python
df = pd.read_csv('Housing.csv')
```

```python
df.shape
```

```python
df.head()
```

```python
numeric_col = df.select_dtypes(include=[np.number]).columns.tolist()
```

```python
target_col = 'price'
```

```python
feature_col = [col for col in numeric_col if col != target_col]
```

```python
X = df[feature_col]
y = df[target_col]
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
coef_df = pd.DataFrame({
    "Feature": feature_col,
    "Coefficient": model.coef_
}).sort_values(by="Coefficient", ascending=False)
```

```python
print("\nModel Coefficients:\n", coef_df.to_string(index=False))
print("Intercept:", model.intercept_)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)
```

```python
print("\n----- Model Evaluation -----")
print(f"MAE  (Mean Absolute Error)     : {mae:.4f}")
print(f"MSE  (Mean Squared Error)      : {mse:.4f}")
print(f"RMSE (Root Mean Squared Error) : {rmse:.4f}")
print(f"R2 Score                       : {r2:.4f}")
```

```python
results_df = X_test.copy()
results_df["Actual_Price"] = y_test.values
results_df["Predicted_Price"] = np.round(y_pred, 2)
results_df["Error"] = np.round(results_df["Actual_Price"] - results_df["Predicted_Price"], 2)
```

```python
results_df.to_csv("prediction_results.csv", index=False)
```

```python
with open("evaluation_metrics.txt", "w") as f:
    f.write("Model Evaluation Metrics\n")
    f.write("========================\n")
    f.write(f"MAE  : {mae:.4f}\n")
    f.write(f"MSE  : {mse:.4f}\n")
    f.write(f"RMSE : {rmse:.4f}\n")
    f.write(f"R2   : {r2:.4f}\n")
```

```python
print("\nPrediction results saved to 'prediction_results.csv'")
print("Evaluation metrics saved to 'evaluation_metrics.txt'")
```
