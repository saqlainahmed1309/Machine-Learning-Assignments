```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/creditcard.csv')
```

```python
data.head()
```

```python
data.isna().sum()
```

```python
data.duplicated().sum()
```

```python
data.drop_duplicates(inplace=True)
```

```python
data.info()
```

```python
X = data.drop(['Class'],axis=1)
y = data['Class']
```

```python
y.value_counts()
```

```python
data.corr()
```

```python
from imblearn.over_sampling import SMOTE
smote = SMOTE(sampling_strategy='minority')
X,y = smote.fit_resample(X,y)
y.value_counts()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=.2, random_state=42)
```

```python
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
from sklearn.preprocessing import RobustScaler
rc = RobustScaler()
X_train[['Amount','Time']] = rc.fit_transform(X_train[['Amount','Time']])
X_test[['Amount','Time']] = rc.transform(X_test[['Amount','Time']])
```

```python
X_train.head()
```

```python
data.corr()
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras import layers
```

```python
model = keras.Sequential([
    layers.Input(shape=(X_train.shape[1],)),
    layers.Dense(64, activation='relu'),
    layers.Dense(32, activation='relu'),
    layers.Dense(16, activation='relu'),
    layers.Dense(1, activation='sigmoid')
])
```

```python
model.summary()
```

```python
model.compile(optimizer='adam',loss='mse',metrics=['r2_score'])
```

```python
model.fit(X_train,y_train,epochs=20,batch_size=64)
```

```python
model.evaluate(X_test,y_test)
```

```python
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score, precision_score, f1_score, recall_score
```

```python
y_pred_prob = model.predict(X_test)
```

```python
y_pred = (y_pred_prob>0.5).astype(int)
```

```python
confusion_matrix(y_test, y_pred)
```

```python
print(classification_report(y_test,y_pred))
```

```python
model_2 = keras.Sequential([
    layers.Input(shape=(X_train.shape[1],)),
    layers.Dense(64, activation='tanh'),
    layers.Dense(32, activation='tanh'),
    layers.Dense(16, activation='tanh'),
    layers.Dense(1, activation='sigmoid')
])
```

```python
model_2.summary()
```

```python
model_2.compile(optimizer='sgd',loss='mse',metrics=['r2_score'])
```

```python
model_2.fit(X_train,y_train,epochs=50,batch_size=128)
```

```python
model.evaluate(X_test,y_test)
```

```python
y_pred_prob = model_2.predict(X_test)
```

```python
y_pred = (y_pred_prob>0.5).astype(int)
```

```python
confusion_matrix(y_test, y_pred)
```

```python
print(classification_report(y_test,y_pred))
```

```python
model_3 = keras.Sequential([
    layers.Input(shape=(X_train.shape[1],)),
    layers.Dense(64, activation='tanh'),
    layers.Dense(32, activation='tanh'),
    layers.Dense(16, activation='tanh'),
    layers.Dense(1, activation='sigmoid')
])
```

```python
model_3.summary()
```

```python
model_3.compile(optimizer='rmsprop',loss='mse',metrics=['r2_score'])
```

```python
model_3.fit(X_train,y_train,epochs=100,batch_size=256)
```

```python
y_pred1 = (model.predict(X_test) > 0.5).astype(int)
y_pred2 = (model_2.predict(X_test) > 0.5).astype(int)
y_pred3 = (model_3.predict(X_test) > 0.5).astype(int)
```

```python
comparison_data = {
    'Model': ['Model 1', 'Model 2', 'Model 3'],
    'Activation':['ReLU', 'Tanh', 'Sigmoid'],
    'Optimizer': ['Adam', 'SGD', 'RMSprop'],
    'Epochs':[20,50,100],
    'Accuracy':[
        round(accuracy_score(y_test, y_pred1), 4),
        round(accuracy_score(y_test, y_pred2), 4),
        round(accuracy_score(y_test, y_pred3), 4)
    ],
    'Precision': [
        round(precision_score(y_test, y_pred1, zero_division=0), 4),
        round(precision_score(y_test, y_pred2, zero_division=0), 4),
        round(precision_score(y_test, y_pred3, zero_division=0), 4)
    ],
    'Recall': [
        round(recall_score(y_test, y_pred1, zero_division=0), 4),
        round(recall_score(y_test, y_pred2, zero_division=0), 4),
        round(recall_score(y_test, y_pred3, zero_division=0), 4)
    ],
    'F1-Score': [
        round(f1_score(y_test, y_pred1, zero_division=0), 4),
        round(f1_score(y_test, y_pred2, zero_division=0), 4),
        round(f1_score(y_test, y_pred3, zero_division=0), 4)
    ]}
```
